## **Part A: Theoretical Fundamentals (Maximum Likelihood & Decision Space)**

### Question 1: MLE Bias & Bessel's Correction

#### (a) Proving that $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ is Biased

**Setup.** Given i.i.d. $\mathbf{X}_1, \dots, \mathbf{X}_n \in \mathbb{R}^m$ with mean $\boldsymbol{\mu}$ and covariance $\boldsymbol{\Sigma}$, the MLE for the covariance matrix is:
$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n}\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

**Step 1: Decompose the centred deviation.** Write $\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n = (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$. Let $\mathbf{d}_i = \mathbf{X}_i - \boldsymbol{\mu}$ and $\bar{\mathbf{d}} = \widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu} = \frac{1}{n}\sum_i \mathbf{d}_i$.

**Step 2: Expand the outer product.**
$$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n \mathbf{d}_i \mathbf{d}_i^T - n\,\bar{\mathbf{d}}\bar{\mathbf{d}}^T$$

**Step 3: Take the expectation term by term.**

$\mathbb{E}[\mathbf{d}_i \mathbf{d}_i^T] = \boldsymbol{\Sigma}$ (by definition), so $\mathbb{E}\!\left[\sum_i \mathbf{d}_i \mathbf{d}_i^T\right] = n\boldsymbol{\Sigma}$.

For the second term: $\mathbb{E}[\bar{\mathbf{d}}\bar{\mathbf{d}}^T] = \text{Var}(\widehat{\boldsymbol{\mu}}_n) = \frac{1}{n}\boldsymbol{\Sigma}$ (since the $\mathbf{X}_i$ are i.i.d.).

**Step 4: Combine.**
$$\mathbb{E}\!\left[\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T\right] = n\boldsymbol{\Sigma} - n \cdot \frac{1}{n}\boldsymbol{\Sigma} = (n-1)\boldsymbol{\Sigma}$$

Therefore:
$$\boxed{\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n}(n-1)\boldsymbol{\Sigma} = \frac{n-1}{n}\boldsymbol{\Sigma}}$$
Since $\frac{n-1}{n} < 1$ for all finite $n$, the MLE systematically **underestimates** every entry of $\boldsymbol{\Sigma}$. The bias magnitude is $-\frac{1}{n}\boldsymbol{\Sigma}$.

#### (b) Bessel's Correction Yields the Unique Unbiased Estimator $\mathbf{S}$

From Step 4 above: $\mathbb{E}\!\left[\sum_i (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T\right] = (n-1)\boldsymbol{\Sigma}$.

Dividing by $(n-1)$ instead of $n$ exactly cancels the bias:
$$\mathbb{E}[\mathbf{S}] = \mathbb{E}\!\left[\frac{1}{n-1}\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T\right] = \frac{(n-1)\boldsymbol{\Sigma}}{n-1} = \boldsymbol{\Sigma} \quad \checkmark$$
The factor $(n-1)$ reflects the **one degree of freedom consumed** in estimating $\boldsymbol{\mu}$ from the data itself.

### Question 2: Type I / Type II Errors & Significance in SHM

#### (a) Definitions in Structural Health Monitoring

Under $H_0$: the asset is **structurally healthy**. Under $H_1$: a fault or damage state is present.

| Error | Symbol | Definition | SHM Consequence |
|---|---|---|---|
| **Type I** | $\alpha$ | $P(\text{reject }H_0 \mid H_0 \text{ true})$ | Raising a **false alarm** — flagging a healthy asset as damaged. Triggers unnecessary plant shutdown. |
| **Type II** | $\beta$ | $P(\text{fail to reject }H_0 \mid H_1 \text{ true})$ | **Missing a real fault** — declaring a damaged asset as healthy. Can lead to catastrophic structural failure. |

The **statistical power** of the test is $1 - \beta$, i.e., the probability of correctly detecting true damage.

#### (b) Effect of Ultra-Conservative $\alpha = 0.0001$ on Power & Confidence Ellipsoid

**Mathematical consequence on power.**  
For a fixed sample size $n$, the Neyman–Pearson framework shows that $\alpha$ and $\beta$ are **inversely linked** through the critical threshold $c_\alpha$. Lowering $\alpha$ raises $c_\alpha$, which shifts the rejection region further into the tail. Formally, if the test statistic $T$ has distribution $F_0$ under $H_0$ and $F_1$ under $H_1$:
$$\text{Power} = 1 - \beta = P(T > c_\alpha \mid H_1) = 1 - F_1(c_\alpha)$$
A much larger $c_\alpha$ (from $\alpha = 0.0001$) drives $F_1(c_\alpha) \to 1$, so $\beta \to 1$ and power $\to 0$.

**Effect on the confidence ellipsoid.**  
For a multivariate Gaussian, the $(1-\alpha)$ confidence ellipsoid is defined by:
$$\mathcal{E}_{1-\alpha} = \left\{\mathbf{x} \in \mathbb{R}^m : (\mathbf{x} - \boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1}(\mathbf{x} - \boldsymbol{\mu}) \leq \chi^2_{m,1-\alpha}\right\}$$
A smaller $\alpha = 0.0001$ pushes the chi-squared quantile $\chi^2_{m,0.9999}$ much higher, **expanding** the ellipsoid volume. In $m$ dimensions the volume scales as:
$$\text{Vol}(\mathcal{E}_{1-\alpha}) \propto \left(\chi^2_{m,1-\alpha}\right)^{m/2} \cdot |\boldsymbol{\Sigma}|^{1/2}$$
A larger ellipsoid accepts more observations as "healthy", dramatically reducing sensitivity to genuine faults — quantifying the power loss geometrically.

## **Part B: Asymptotic Distributions & Slutsky's Theorem**

### Slutsky's Theorem (Rigorous Statement)

Let $\{\mathbf{A}_n\}$ and $\{\mathbf{B}_n\}$ be sequences of random matrices/vectors such that:
$$\mathbf{A}_n \xrightarrow{d} \mathbf{A}, \qquad \mathbf{B}_n \xrightarrow{p} \mathbf{c}$$
where $\mathbf{c}$ is a **deterministic constant** (matrix or vector). Then:
$$\mathbf{B}_n \mathbf{A}_n \xrightarrow{d} \mathbf{c}\mathbf{A}, \qquad \mathbf{A}_n + \mathbf{B}_n \xrightarrow{d} \mathbf{A} + \mathbf{c}$$

### Proof: Substituting $\mathbf{S}$ for $\boldsymbol{\Sigma}$

**Step 1: The CLT baseline.**  
By the Lindeberg–Lévy Multivariate CLT:
$$\mathbf{W}_n := \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$$

**Step 2: Consistency of $\mathbf{S}$.**  
By the Weak Law of Large Numbers applied to the centred products:
$$\mathbf{S} = \frac{1}{n-1}\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \xrightarrow{p} \boldsymbol{\Sigma}$$
Therefore $\mathbf{S}^{-1} \xrightarrow{p} \boldsymbol{\Sigma}^{-1}$ (by the continuous mapping theorem, since inversion is continuous on the set of positive-definite matrices).

**Step 3: Apply Slutsky's Theorem.**  
Consider the standardized statistic:
$$\mathbf{S}^{-1/2}\,\mathbf{W}_n = \mathbf{S}^{-1/2} \cdot \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$$
Since $\mathbf{S}^{-1/2} \xrightarrow{p} \boldsymbol{\Sigma}^{-1/2}$ and $\mathbf{W}_n \xrightarrow{d} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$, Slutsky's theorem gives:
$$\mathbf{S}^{-1/2}\mathbf{W}_n \xrightarrow{d} \boldsymbol{\Sigma}^{-1/2}\mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma}) = \mathscr{N}(\mathbf{0}, \mathbf{I})$$

**Step 4: Operational distribution.**  
Equivalently, for large $n$ we may write:
$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathscr{N}(\mathbf{0}, \mathbf{S})$$
which, re-scaling, operationalizes the practical modelling distribution:
$$\boxed{\widehat{\boldsymbol{\mu}}_n \sim \mathscr{N}\!\left(\boldsymbol{\mu},\, \frac{1}{n}\mathbf{S}\right)}$$
This justifies replacing the latent $\boldsymbol{\Sigma}$ with its consistent estimator $\mathbf{S}$ in all downstream inference without invalidating the Gaussian approximation asymptotically.

## **Part C: Numerical Verification — Wilks' Lambda MANOVA Test**

```python
import numpy as np
import pandas as pd
import scipy.stats as stats

# ── DO NOT MODIFY: Synthetic Strain Sensor Data Generator ────────────────────
np.random.seed(42)
n_samples  = 5000
n_features = 3

base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)

# Injecting subtle hidden structural drift in the final 1,000 snapshots
base_data[4000:, 0] += 0.015   # Sensor 1 drift
base_data[4000:, 2] -= 0.010   # Sensor 3 drift

df_strain = pd.DataFrame(base_data, columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'])
# ─────────────────────────────────────────────────────────────────────────────


def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:
    """
    Partitions the dataset into g consecutive, non-overlapping blocks and
    evaluates first-moment homogeneity via Wilks' Lambda and Bartlett's
    Chi-Square asymptotic transformation.

    Parameters
    ----------
    df       : pd.DataFrame  — raw sensor observations  (n × m)
    g_chunks : int           — number of equal partitions

    Returns
    -------
    dict with keys: wilks_lambda, chi2_stat, p_value, df_chi2, conclusion
    """
    X     = df.values          # (n × m) array
    n, m  = X.shape
    g     = g_chunks
    chunk = n // g             # observations per block  (n_j)

    # ── 1. Split into g consecutive equal blocks ─────────────────────────────
    chunks = [X[j * chunk : (j + 1) * chunk] for j in range(g)]

    # ── 2. Global mean and per-chunk means ───────────────────────────────────
    grand_mean  = X[:g * chunk].mean(axis=0)         # (m,)
    chunk_means = np.array([c.mean(axis=0) for c in chunks])  # (g × m)

    # ── 3a. Within-chunk scatter matrix W ────────────────────────────────────
    #   W = Σ_j Σ_i (x_ij − x̄_j)(x_ij − x̄_j)^T
    W = np.zeros((m, m))
    for j, c in enumerate(chunks):
        diff = c - chunk_means[j]           # (n_j × m)
        W   += diff.T @ diff                # accumulate outer products

    # ── 3b. Between-chunk scatter matrix B ───────────────────────────────────
    #   B = Σ_j n_j (x̄_j − x̄)(x̄_j − x̄)^T
    B = np.zeros((m, m))
    for j in range(g):
        dev  = (chunk_means[j] - grand_mean).reshape(-1, 1)   # (m × 1)
        B   += chunk * (dev @ dev.T)

    # ── 4. Wilks' Lambda ─────────────────────────────────────────────────────
    #   Λ = |W| / |B + W|
    det_W   = np.linalg.det(W)
    det_BW  = np.linalg.det(B + W)
    Lambda  = det_W / det_BW

    # ── 5. Bartlett's Chi-Square approximation ───────────────────────────────
    #   χ² = −(n − 1 − (m + g)/2) · ln Λ
    scale   = n - 1 - (m + g) / 2
    chi2    = -scale * np.log(Lambda)

    # ── 6. Degrees of freedom = m(g − 1); p-value ────────────────────────────
    dof     = m * (g - 1)
    p_value = 1 - stats.chi2.cdf(chi2, df=dof)

    # ── 7. Decision ──────────────────────────────────────────────────────────
    alpha   = 0.05
    if p_value < alpha:
        conclusion = (f"REJECT H₀ (p = {p_value:.4e} < α = {alpha}). "
                      "Evidence of a mean shift — the first moment is NOT homogeneous.")
    else:
        conclusion = (f"FAIL TO REJECT H₀ (p = {p_value:.4f} ≥ α = {alpha}). "
                      "No significant mean shift detected — first moment is homogeneous.")

    return dict(
        wilks_lambda = Lambda,
        chi2_stat    = chi2,
        df_chi2      = dof,
        p_value      = p_value,
        conclusion   = conclusion
    )


# ── Execute ───────────────────────────────────────────────────────────────────
results = verify_first_moment_homogeneity(df_strain, g_chunks=5)

print("══════════════════════════════════════════════════════")
print("  MANOVA First-Moment Homogeneity Test (Wilks' Λ)")
print("══════════════════════════════════════════════════════")
print(f"  Wilks' Lambda  (Λ) : {results['wilks_lambda']:.6f}")
print(f"  Bartlett χ²        : {results['chi2_stat']:.4f}")
print(f"  Degrees of freedom : {results['df_chi2']}")
print(f"  p-value            : {results['p_value']:.4e}")
print("──────────────────────────────────────────────────────")
print(f"  Conclusion: {results['conclusion']}")
print("══════════════════════════════════════════════════════")
```

### Interpretation

The injected drifts (`+0.015` on Sensor 1 and `−0.010` on Sensor 3 in the final block) are small individually but statistically detectable because Wilks' Lambda is extremely sensitive to **joint mean displacement** across all channels simultaneously. The chi-squared test operates on $m(g-1) = 3 \times 4 = 12$ degrees of freedom, reflecting $g-1=4$ independent group contrasts in $m=3$ dimensions. A $p$-value below $0.05$ confirms the baseline centre has shifted over time — the i.i.d. mean-homogeneity assumption is violated and a global PCA/FA layer should **not** be initialised until the drift is removed or modelled.

# **Geometric Subspace Optimization via PCA**

## **Part A: Theoretical Fundamentals**

### Question 1: Covariance of PCA-Transformed Coordinates

#### (a) Proof that $\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{\Lambda}$

Given the spectral decomposition $\boldsymbol{\Sigma} = \mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T$ with $\mathbf{P}^T\mathbf{P} = \mathbf{I}$ (orthonormal columns), and the projection $\mathbf{Z}_i = \mathbf{P}^T \widetilde{\mathbf{X}}_i$:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}[(\mathbf{P}^T \widetilde{\mathbf{X}}_i)(\mathbf{P}^T \widetilde{\mathbf{X}}_i)^T]$$
$$= \mathbb{E}[\mathbf{P}^T \widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T \mathbf{P}]$$
$$= \mathbf{P}^T \,\mathbb{E}[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T]\, \mathbf{P}$$
$$= \mathbf{P}^T \boldsymbol{\Sigma} \mathbf{P} = \mathbf{P}^T (\mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T) \mathbf{P}$$
$$= (\mathbf{P}^T \mathbf{P})\,\boldsymbol{\Lambda}\,(\mathbf{P}^T \mathbf{P}) = \mathbf{I}\,\boldsymbol{\Lambda}\,\mathbf{I} = \boxed{\boldsymbol{\Lambda}}$$

#### (b) Geometric & Statistical Meaning

The diagonal structure of $\boldsymbol{\Lambda}$ means $\text{Cov}(Z_{i,j}, Z_{i,k}) = 0$ for all $j \neq k$. Geometrically, the coordinate axes of $\mathbf{Z}$ align with the **principal axes of the data ellipsoid** — each axis captures a unique, independent mode of variation. Statistically, distinct principal coordinates are **uncorrelated** (and Gaussian if the data are Gaussian, which implies **independence**), making downstream modelling far simpler.

---

### Question 2: Total Variance Conservation

#### (a) Proof via Trace Invariance

Using the cyclic property of the trace: $\text{tr}(\mathbf{ABC}) = \text{tr}(\mathbf{BCA}) = \text{tr}(\mathbf{CAB})$,
$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T) = \text{tr}(\boldsymbol{\Lambda}\mathbf{P}^T\mathbf{P}) = \text{tr}(\boldsymbol{\Lambda}\cdot\mathbf{I}) = \text{tr}(\boldsymbol{\Lambda}) = \sum_{j=1}^m \lambda_j \quad \checkmark$$

#### (b) Explained & Residual Variance Ratios

$$\Phi(k) = \frac{\sum_{j=1}^k \widehat{\lambda}_j}{\sum_{j=1}^m \widehat{\lambda}_j} \qquad \text{(Cumulative Explained Variance Ratio)}$$
$$\Psi(k) = 1 - \Phi(k) = \frac{\sum_{j=k+1}^m \widehat{\lambda}_j}{\sum_{j=1}^m \widehat{\lambda}_j} \qquad \text{(Residual Unexplained Variance Ratio)}$$

---

### Question 3: Truncation, Reconstruction & Diagnostic Statistics

#### (a) Reconstructed & Residual Vectors

$$\widehat{\mathbf{x}}_i = \widehat{\boldsymbol{\mu}}_n + \widehat{\mathbf{P}}_k \mathbf{z}_{i,k}$$
$$\mathbf{e}_i = (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n) - \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} = \widehat{\mathbf{P}}_{m-k}\mathbf{z}_{i,m-k}$$

#### (b) Pythagorean Decomposition

Since $\widehat{\mathbf{P}}_k$ and $\widehat{\mathbf{P}}_{m-k}$ are **mutually orthogonal** (their columns span complementary subspaces), the cross term vanishes:
$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \|\widehat{\mathbf{P}}_k\mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k}\mathbf{z}_{i,m-k}\|^2$$
$$= \mathbf{z}_{i,k}^T(\widehat{\mathbf{P}}_k^T\widehat{\mathbf{P}}_k)\mathbf{z}_{i,k} + 2\mathbf{z}_{i,k}^T\underbrace{(\widehat{\mathbf{P}}_k^T\widehat{\mathbf{P}}_{m-k})}_{=\mathbf{0}}\mathbf{z}_{i,m-k} + \mathbf{z}_{i,m-k}^T(\widehat{\mathbf{P}}_{m-k}^T\widehat{\mathbf{P}}_{m-k})\mathbf{z}_{i,m-k}$$
$$= \|\mathbf{z}_{i,k}\|^2 + \|\mathbf{z}_{i,m-k}\|^2 \quad \checkmark$$

#### (c) Hotelling's $T^2$ vs. $Q$ — Diagnostic Assignment

| Statistic | Formula | Monitors | Sensitive To |
|---|---|---|---|
| Hotelling's $T^2$ | $T^2_i = \sum_{j=1}^k z_{i,j}^2 / \widehat{\lambda}_j$ | **Within** the $k$-dim principal subspace | Amplitude/direction changes in known structural modes |
| $Q$ (SPE) | $Q_i = \sum_{j=k+1}^m z_{i,j}^2$ | **Residual** space outside the subspace | New modes of variation not seen in baseline |

- **Extreme wind loads** → energise existing structural modes → $T^2$ is most sensitive (loading stays in the principal subspace but at unusual magnitude).
- **Fatigue crack** → introduces a new load pathway, changing the covariance structure itself → $Q$ (SPE) is most sensitive (energy leaks into the residual space).

## **Part B: PCA Numerical Verification**

```python
import numpy as np

# ── Synthetic data (4 sensors, 3 000 baseline snapshots) ─────────────────────
np.random.seed(0)
n, m = 3000, 4

true_cov = np.array([
    [1.00, 0.80, 0.60, 0.10],
    [0.80, 1.00, 0.70, 0.05],
    [0.60, 0.70, 1.00, 0.08],
    [0.10, 0.05, 0.08, 1.50],
])
X_raw = np.random.multivariate_normal(mean=np.zeros(m), cov=true_cov, size=n)


class PCADiagnosticEngine:
    """
    Computes Hotelling T² and Q statistics for every subspace cutoff k.
    """

    def fit(self, X: np.ndarray) -> dict:
        """
        Parameters
        ----------
        X : np.ndarray  shape (n, m)

        Returns
        -------
        dict with eigenvalues, eigenvectors, per-k T² array, per-k Q array,
        mean T² and mean Q per k.
        """
        n_obs, m_feat = X.shape

        # ── Step 1: Centre (standardise) ─────────────────────────────────────
        mu_hat = X.mean(axis=0)                # (m,)
        X_c    = X - mu_hat                    # (n × m) zero-mean

        # ── Step 2: Unbiased sample covariance → eigendecomposition ──────────
        S = (X_c.T @ X_c) / (n_obs - 1)       # (m × m)  ddof=1
        eigvals, eigvecs = np.linalg.eigh(S)   # ascending order from eigh

        # Sort descending  λ₁ ≥ λ₂ ≥ … ≥ λ_m
        idx     = np.argsort(eigvals)[::-1]
        eigvals = eigvals[idx]                 # (m,)
        P       = eigvecs[:, idx]              # (m × m)  — columns = eigenvectors

        # ── Step 3: Project all snapshots into full PC space ─────────────────
        Z = X_c @ P                            # (n × m)  — all PC scores

        # ── Step 4: Loop over subspace cutoffs k ∈ {1, 2, 3} ─────────────────
        k_values   = list(range(1, m_feat))    # [1, 2, 3] for m=4
        T2_all     = {}                        # k → (n,) array
        Q_all      = {}                        # k → (n,) array
        mean_T2    = {}
        mean_Q     = {}

        for k in k_values:
            Z_k    = Z[:, :k]                  # (n × k)  in-subspace scores
            Z_res  = Z[:, k:]                  # (n × m-k) residual scores
            lam_k  = eigvals[:k]               # (k,)     kept eigenvalues

            # Hotelling T²_i = Σ_{j=1}^k  z_{i,j}² / λ_j
            T2_i   = np.sum((Z_k ** 2) / lam_k, axis=1)   # (n,)

            # Q_i (SPE) = Σ_{j=k+1}^m  z_{i,j}²
            Q_i    = np.sum(Z_res ** 2, axis=1)             # (n,)

            T2_all[k]  = T2_i
            Q_all[k]   = Q_i
            mean_T2[k] = T2_i.mean()
            mean_Q[k]  = Q_i.mean()

        # ── Step 5: Report ────────────────────────────────────────────────────
        print("══════════════════════════════════════════════════════")
        print("  PCA Diagnostic Engine — T² & Q per Subspace Size k")
        print("══════════════════════════════════════════════════════")
        print(f"  Eigenvalues (descending): {np.round(eigvals, 4)}")
        total_var = eigvals.sum()
        for k in k_values:
            expl = eigvals[:k].sum() / total_var * 100
            print(f"  k={k}: mean T² = {mean_T2[k]:.4f} | "
                  f"mean Q = {mean_Q[k]:.4f} | "
                  f"Variance explained = {expl:.1f}%")
        print("══════════════════════════════════════════════════════")

        return dict(
            eigenvalues = eigvals,
            eigenvectors = P,
            T2_arrays   = T2_all,
            Q_arrays    = Q_all,
            mean_T2     = mean_T2,
            mean_Q      = mean_Q,
        )


engine  = PCADiagnosticEngine()
pca_res = engine.fit(X_raw)
```

# **Latent Subspace Decomposition via Factor Analysis (FA)**

## **Part A: Theoretical Fundamentals**

### Question 1: Fundamental Equation of Factor Analysis

#### Proof that $\mathbf{R} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$

The generative model: $\mathbf{Z}_i = \boldsymbol{\Lambda}\mathbf{F}_i + \boldsymbol{\epsilon}_i$, with $\mathbf{F}_i \perp \boldsymbol{\epsilon}_i$, $\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T]=\mathbf{I}$, $\mathbb{E}[\boldsymbol{\epsilon}_i\boldsymbol{\epsilon}_i^T]=\boldsymbol{\Psi}$.

$$\mathbf{R} = \mathbb{E}[\mathbf{Z}_i\mathbf{Z}_i^T] = \mathbb{E}[(\boldsymbol{\Lambda}\mathbf{F}_i + \boldsymbol{\epsilon}_i)(\boldsymbol{\Lambda}\mathbf{F}_i + \boldsymbol{\epsilon}_i)^T]$$
$$= \boldsymbol{\Lambda}\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T]\boldsymbol{\Lambda}^T + \boldsymbol{\Lambda}\underbrace{\mathbb{E}[\mathbf{F}_i\boldsymbol{\epsilon}_i^T]}_{=\mathbf{0}} + \underbrace{\mathbb{E}[\boldsymbol{\epsilon}_i\mathbf{F}_i^T]}_{=\mathbf{0}}\boldsymbol{\Lambda}^T + \mathbb{E}[\boldsymbol{\epsilon}_i\boldsymbol{\epsilon}_i^T]$$
$$= \boldsymbol{\Lambda}\mathbf{I}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi} = \boxed{\boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}}$$

The $j$-th diagonal entry gives the variance decomposition for sensor $j$:
$$1 = h_j^2 + \varphi_j^2$$
where $h_j^2 = \sum_{r=1}^k \lambda_{j,r}^2$ (**communality**) = shared variance captured by the $k$ common factors, and $\varphi_j^2 = \boldsymbol{\Psi}_{jj}$ (**uniqueness**) = isolated sensor noise. In diagnostics, a near-zero uniqueness indicates a sensor dominated by global structural load; a near-unity uniqueness (e.g., > 0.95) flags a hardware malfunction or instrumentation noise source.

---

### Question 2: Varimax Rotation

- **Raw PCA loadings** are constrained to lie along the eigenvectors of $\mathbf{S}$, ordered by variance magnitude. Each eigenvector typically loads on *many* sensors at once ("mixed" structure).
- **Varimax-rotated FA loadings** apply an orthogonal rotation $\mathbf{T}$ (maximising the variance of squared loadings per factor) that redistributes the loading energy so each factor loads *heavily on a few sensors and near-zero on the rest* — "simple structure".

Rotation preserves communalities because:
$$h_j^2 = \|\boldsymbol{\lambda}_j^T \mathbf{T}\|^2 = \|\boldsymbol{\lambda}_j\|^2$$
since $\mathbf{T}$ is orthogonal ($\mathbf{T}^T\mathbf{T}=\mathbf{I}$). Similarly, $\boldsymbol{\Lambda}_{\text{rot}}\boldsymbol{\Lambda}_{\text{rot}}^T = \boldsymbol{\Lambda}\mathbf{T}\mathbf{T}^T\boldsymbol{\Lambda}^T = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T$, leaving the covariance approximation unchanged.

---

### Thomson's Regression Score Estimator

#### Proof that $\mathbf{f}_i = \boldsymbol{\Lambda}^T\mathbf{R}^{-1}\mathbf{z}_i$

Construct the joint vector $\mathbf{Y}_i = [\mathbf{Z}_i^T, \mathbf{F}_i^T]^T$. Its covariance matrix:

$$\boldsymbol{\Sigma}_{YY} = \begin{bmatrix} \mathbb{E}[\mathbf{Z}_i\mathbf{Z}_i^T] & \mathbb{E}[\mathbf{Z}_i\mathbf{F}_i^T] \\ \mathbb{E}[\mathbf{F}_i\mathbf{Z}_i^T] & \mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] \end{bmatrix}$$

$\mathbb{E}[\mathbf{Z}_i\mathbf{F}_i^T] = \mathbb{E}[(\boldsymbol{\Lambda}\mathbf{F}_i + \boldsymbol{\epsilon}_i)\mathbf{F}_i^T] = \boldsymbol{\Lambda}\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i\mathbf{F}_i^T] = \boldsymbol{\Lambda}$

$$\boldsymbol{\Sigma}_{YY} = \begin{bmatrix} \mathbf{R} & \boldsymbol{\Lambda} \\ \boldsymbol{\Lambda}^T & \mathbf{I} \end{bmatrix} \quad \checkmark$$

Applying the MVN conditional mean formula with $\boldsymbol{\mu}_1 = \boldsymbol{\mu}_2 = \mathbf{0}$:
$$\mathbf{f}_i = \mathbb{E}[\mathbf{F}_i | \mathbf{z}_i] = \mathbf{0} + \boldsymbol{\Sigma}_{21}\boldsymbol{\Sigma}_{11}^{-1}(\mathbf{z}_i - \mathbf{0}) = \boldsymbol{\Lambda}^T \mathbf{R}^{-1} \mathbf{z}_i \quad \checkmark$$

## **Part B: FA Engine & 2×2 Diagnostic Dashboard**

```python
import numpy as np
import pandas as pd
from sklearn.decomposition import FactorAnalysis
from scipy.linalg import solve
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ── Synthetic dataset (re-use Part C data or generate fresh) ─────────────────
np.random.seed(42)
n_s = 5000
base_data2 = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09,0.02,0.01],[0.02,0.06,0.03],[0.01,0.03,0.05]],
    size=n_s
)
df_fa = pd.DataFrame(base_data2, columns=['Sensor_1','Sensor_2','Sensor_3'])


def run_fa_engine(df: pd.DataFrame, k: int = 2) -> dict:
    """
    Full Factor Analysis engine with Varimax rotation and Thomson scoring.

    Parameters
    ----------
    df : pd.DataFrame — raw sensor observations
    k  : int          — number of latent factors to extract

    Returns
    -------
    dict with all decomposition artefacts
    """
    # ── Layer 1: Z-Score Mapping with zero-variance guard ────────────────────
    numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
    X_raw = df[numeric_cols].values          # (n × m)
    n_obs, m_feat = X_raw.shape

    mu  = X_raw.mean(axis=0)
    std = X_raw.std(axis=0, ddof=1)
    std = np.where(std < 1e-15, 1e-15, std)  # zero-variance guardrail
    Z   = (X_raw - mu) / std                 # (n × m) standard normal space

    # ── Layer 2: Dimensionality constraint checks ─────────────────────────────
    if n_obs <= m_feat:
        raise ValueError(f"Need n > m for reliable FA. Got n={n_obs}, m={m_feat}.")
    if k >= m_feat:
        raise ValueError(f"Latent dimension k must be < m. Got k={k}, m={m_feat}.")

    # ── Layer 3: Factor Analysis with Varimax rotation ───────────────────────
    fa = FactorAnalysis(n_components=k, rotation='varimax', random_state=42)
    fa.fit(Z)

    Lambda_rot = fa.components_.T              # (m × k)  — rotated loadings
    Psi        = fa.noise_variance_            # (m,)     — uniquenesses φ²
    h2         = 1.0 - Psi                     # (m,)     — communalities  h²

    # ── Thomson's MMSE factor scores: F = Z R⁻¹ Λ_rot ───────────────────────
    R   = np.corrcoef(Z, rowvar=False)         # empirical correlation matrix
    R_inv = np.linalg.inv(R)
    F   = Z @ R_inv @ Lambda_rot               # (n × k)

    # ── Performance diagnostics ──────────────────────────────────────────────
    factor_var = F.var(axis=0, ddof=1)         # (k,)
    avg_comm   = h2.mean() * 100
    avg_unique = Psi.mean() * 100

    print("══════════════════════════════════════════════════════")
    print("  Factor Analysis Engine — Decomposition Summary")
    print("══════════════════════════════════════════════════════")
    for j, col in enumerate(numeric_cols):
        print(f"  {col:<12}  h² = {h2[j]:.4f}   φ² = {Psi[j]:.4f}")
    print(f"  {'─'*50}")
    print(f"  Avg System Communality  : {avg_comm:.2f}%")
    print(f"  Avg System Uniqueness   : {avg_unique:.2f}%")
    for r in range(k):
        print(f"  Factor {r+1} score variance: {factor_var[r]:.4f}")
    print("══════════════════════════════════════════════════════")

    return dict(
        sensor_names  = numeric_cols,
        Z             = Z,
        Lambda_rot    = Lambda_rot,
        Psi           = Psi,
        h2            = h2,
        F             = F,
        factor_var    = factor_var,
    )


fa_res = run_fa_engine(df_fa, k=2)


# ── 2×2 Diagnostic Dashboard ─────────────────────────────────────────────────
def plot_fa_dashboard(res: dict):
    sensors      = res['sensor_names']
    Lambda_rot   = res['Lambda_rot']         # (m × k)
    h2           = res['h2']
    Psi          = res['Psi']
    factor_var   = res['factor_var']
    k            = Lambda_rot.shape[1]
    factor_labels = [f'Factor {r+1}' for r in range(k)]

    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=[
            'Structural Loadings Matrix |Λ|',
            'Variance Partitioning (h² vs φ²)',
            'Sensor Uniqueness Noise Floor (φ²)',
            'Latent Factor Scores Empirical Variance'
        ],
        horizontal_spacing=0.24,
        vertical_spacing=0.28,
    )

    # ── (1,1): Heatmap of |Λ_rot| ────────────────────────────────────────────
    fig.add_trace(go.Heatmap(
        z          = np.abs(Lambda_rot),
        x          = factor_labels,
        y          = sensors,
        colorscale = 'YlOrRd',
        colorbar   = dict(
            x=−0.15, y=0.78, len=0.38,
            yanchor='middle', xanchor='right',
            title='|λ|'
        ),
        showscale=True,
        name='Loadings'
    ), row=1, col=1)

    # ── (1,2): Stacked horizontal bar  h² & φ² ───────────────────────────────
    fig.add_trace(go.Bar(
        y           = sensors,
        x           = h2 * 100,
        orientation = 'h',
        name        = 'Communality h²',
        marker_color= '#1f77b4',
    ), row=1, col=2)
    fig.add_trace(go.Bar(
        y           = sensors,
        x           = Psi * 100,
        orientation = 'h',
        name        = 'Uniqueness φ²',
        marker_color= '#ff7f0e',
    ), row=1, col=2)
    fig.update_xaxes(range=[0, 100], row=1, col=2)

    # ── (2,1): Uniqueness line profile ───────────────────────────────────────
    fig.add_trace(go.Scatter(
        x    = sensors,
        y    = Psi,
        mode = 'lines+markers',
        name = 'Uniqueness φ²',
        line = dict(color='#d62728', width=2, dash='dashdot'),
        marker=dict(symbol='x', size=10)
    ), row=2, col=1)
    fig.update_xaxes(tickangle=25, row=2, col=1)

    # ── (2,2): Factor variance bar chart ─────────────────────────────────────
    fig.add_trace(go.Bar(
        x    = factor_labels,
        y    = factor_var,
        name = 'Factor Score Variance',
        marker=dict(
            color      = '#2ca02c',
            line       = dict(color='black', width=0.5)
        )
    ), row=2, col=2)

    fig.update_layout(
        title_text  = 'FA 2×2 Diagnostic Dashboard',
        width=1250, height=750,
        template    = 'plotly_white',
        barmode     = 'stack',
        legend      = dict(orientation='h', x=0.5, y=1.02, xanchor='center'),
        margin      = dict(t=150, b=60, l=140, r=80),
    )
    fig.show()


plot_fa_dashboard(fa_res)
```

```python
import numpy as np
import pandas as pd
from sklearn.decomposition import FactorAnalysis
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ═══════════════════════════════════════════════════════════════════
#  HELPER FUNCTIONS
# ═══════════════════════════════════════════════════════════════════

def compute_pca(X: np.ndarray):
    """Centre, compute S, eigendecompose. Returns (mu, S, eigvals, P, Z_all)."""
    n, m   = X.shape
    mu     = X.mean(axis=0)
    X_c    = X - mu
    S      = (X_c.T @ X_c) / (n - 1)
    ev, P  = np.linalg.eigh(S)
    idx    = np.argsort(ev)[::-1]
    ev, P  = ev[idx], P[:, idx]
    Z_all  = X_c @ P                           # (n × m) full score matrix
    return mu, S, ev, P, Z_all


def pca_t2_q(Z_all, eigvals, k):
    lam_k  = eigvals[:k]
    T2_i   = np.sum((Z_all[:, :k] ** 2) / lam_k, axis=1)
    Q_i    = np.sum(Z_all[:, k:] ** 2, axis=1)
    return T2_i.mean(), Q_i.mean()


def compute_fa(X: np.ndarray, k: int):
    """Z-score, fit FA with Varimax, return loadings / h² / φ² / F."""
    n, m   = X.shape
    std    = X.std(axis=0, ddof=1)
    std    = np.where(std < 1e-15, 1e-15, std)
    Z      = (X - X.mean(axis=0)) / std
    fa     = FactorAnalysis(n_components=k, rotation='varimax', random_state=42)
    fa.fit(Z)
    L_rot  = fa.components_.T                  # (m × k)
    Psi    = fa.noise_variance_                 # (m,)
    h2     = 1.0 - Psi
    R      = np.corrcoef(Z, rowvar=False)
    F      = Z @ np.linalg.inv(R) @ L_rot      # (n × k)  Thomson scores
    return Z, L_rot, h2, Psi, F


# ═══════════════════════════════════════════════════════════════════
#  Compute everything
# ═══════════════════════════════════════════════════════════════════
X       = df_asset.values
sensors = df_asset.columns.tolist()
n, m    = X.shape
k_fa    = 2                                    # latent factors for FA

mu, S, eigvals, P, Z_all = compute_pca(X)
total_var = eigvals.sum()

expl_pct   = eigvals / total_var * 100                   # marginal explained %
cum_pct    = np.cumsum(expl_pct)                         # cumulative
resid_pct  = 100 - cum_pct

k_range    = list(range(1, m))
mean_T2    = [pca_t2_q(Z_all, eigvals, k)[0] for k in k_range]
mean_Q     = [pca_t2_q(Z_all, eigvals, k)[1] for k in k_range]

Z_fa, L_rot, h2, Psi, F_scores = compute_fa(X, k=k_fa)
factor_var = F_scores.var(axis=0, ddof=1)
pc_labels  = [f'PC {j+1}' for j in range(m)]
f_labels   = [f'Factor {r+1}' for r in range(k_fa)]


# ═══════════════════════════════════════════════════════════════════
#  UI LAYOUT 1 — PCA Optimization Suite (2 × 3)
# ═══════════════════════════════════════════════════════════════════
fig_pca = make_subplots(
    rows=2, cols=3,
    subplot_titles=[
        'Feature Loadings |P_jr|',
        'Eigenvalues (λⱼ)',
        'Marginal & Cumulative Variance',
        'Residual Unexplained Variance %',
        'Mean Hotelling T² vs k',
        'Mean Q Statistic (SPE) vs k',
    ],
    horizontal_spacing=0.12,
    vertical_spacing=0.26,
)

# Panel (1,1) — Loadings heatmap
fig_pca.add_trace(go.Heatmap(
    z=np.abs(P), x=pc_labels, y=sensors,
    colorscale='Blues', showscale=True,
    colorbar=dict(len=0.4, x=0.28, y=0.78, title='|P|')
), row=1, col=1)

# Panel (1,2) — Eigenvalue bar
fig_pca.add_trace(go.Bar(
    x=pc_labels, y=eigvals,
    marker_color='#4472C4', name='Eigenvalues'
), row=1, col=2)

# Panel (1,3) — Marginal bars + cumulative line
fig_pca.add_trace(go.Bar(
    x=pc_labels, y=expl_pct,
    marker_color='#70AD47', name='Marginal Var %'
), row=1, col=3)
fig_pca.add_trace(go.Scatter(
    x=pc_labels, y=cum_pct,
    mode='lines+markers', name='Cumulative Var %',
    line=dict(dash='dash', color='#C00000', width=2)
), row=1, col=3)

# Panel (2,1) — Residual unexplained %
fig_pca.add_trace(go.Bar(
    x=pc_labels, y=resid_pct,
    marker_color='#ED7D31', name='Residual Var %'
), row=2, col=1)

# Panel (2,2) — Mean T² vs k
fig_pca.add_trace(go.Scatter(
    x=[f'k={k}' for k in k_range], y=mean_T2,
    mode='lines+markers', name='Mean T²',
    line=dict(color='#7030A0', width=2),
    marker=dict(size=9)
), row=2, col=2)

# Panel (2,3) — Mean Q vs k
fig_pca.add_trace(go.Scatter(
    x=[f'k={k}' for k in k_range], y=mean_Q,
    mode='lines+markers', name='Mean Q (SPE)',
    line=dict(color='#C00000', width=2),
    marker=dict(symbol='diamond', size=9)
), row=2, col=3)

fig_pca.update_layout(
    title_text='PCA Optimization Suite — 4-Sensor Asset',
    width=1350, height=780,
    template='plotly_white',
    showlegend=True,
    legend=dict(orientation='h', x=0.5, y=1.06, xanchor='center'),
    margin=dict(t=160, b=60, l=80, r=60)
)
fig_pca.show()


# ═══════════════════════════════════════════════════════════════════
#  UI LAYOUT 2 — FA Latent Subspace Suite (2 × 2)
# ═══════════════════════════════════════════════════════════════════
fig_fa = make_subplots(
    rows=2, cols=2,
    subplot_titles=[
        'Structural Loadings Matrix |Λ|',
        'Variance Allocation (h² vs φ²)',
        'Sensor Uniqueness Noise Floor (φ²)',
        'Latent Factor Score Empirical Variance',
    ],
    horizontal_spacing=0.24,
    vertical_spacing=0.28,
)

# Panel (1,1) — FA loadings heatmap
fig_fa.add_trace(go.Heatmap(
    z=np.abs(L_rot), x=f_labels, y=sensors,
    colorscale='YlOrRd', showscale=True,
    colorbar=dict(x=-0.15, y=0.78, len=0.38,
                  yanchor='middle', xanchor='right', title='|λ|')
), row=1, col=1)

# Panel (1,2) — Stacked horizontal bar h² & φ²
fig_fa.add_trace(go.Bar(
    y=sensors, x=h2*100, orientation='h',
    name='Communality h²', marker_color='#1f77b4'
), row=1, col=2)
fig_fa.add_trace(go.Bar(
    y=sensors, x=Psi*100, orientation='h',
    name='Uniqueness φ²', marker_color='#ff7f0e'
), row=1, col=2)
fig_fa.update_xaxes(range=[0,100], row=1, col=2)

# Panel (2,1) — Uniqueness noise floor line
fig_fa.add_trace(go.Scatter(
    x=sensors, y=Psi, mode='lines+markers',
    name='Uniqueness φ²',
    line=dict(color='#d62728', width=2, dash='dashdot'),
    marker=dict(symbol='x', size=10)
), row=2, col=1)
fig_fa.update_xaxes(tickangle=25, row=2, col=1)

# Panel (2,2) — Factor score variance
fig_fa.add_trace(go.Bar(
    x=f_labels, y=factor_var,
    name='Factor Variance',
    marker=dict(color='#2ca02c', line=dict(color='black', width=0.5))
), row=2, col=2)

fig_fa.update_layout(
    title_text='FA Latent Subspace Suite — 4-Sensor Asset',
    width=1250, height=750,
    template='plotly_white',
    barmode='stack',
    legend=dict(orientation='h', x=0.5, y=1.02, xanchor='center'),
    margin=dict(t=150, b=60, l=140, r=80)
)
fig_fa.show()
```

## **Part B: Analysis Questions**

### Question 1: The Total Variance Illusion

**1. What does a Uniqueness $\varphi^2 \approx 1$ tell us about Sensor 4?**

A uniqueness near 100% means that virtually **none** of Sensor 4's variance is shared with the other sensors through common latent factors. Its signal is dominated by localised, idiosyncratic noise — in practice this flags an **instrumentation fault** (electrical short, sensor miscalibration, loose connection) rather than any structural load effect. The factor model correctly quarantines it in the noise floor $\boldsymbol{\Psi}$.

**2. How does Sensor 4's noise inflate PCA eigenvalues?**

PCA maximises total variance $\text{tr}(\mathbf{S})$ without distinguishing *type* of variance. Because $\text{Var}(\text{Sensor 4}) \approx 2.0$ (roughly double that of structural sensors), its contribution to every eigenvalue is large. The leading principal components are "pulled" toward Sensor 4's noise axis, artificially inflating $\lambda_1$ and $\lambda_2$. The PCA engine interprets random white noise as a dominant mode of the structure — a false signal.

**3. Why is PCA-only monitoring risky?**

An automated plant anomaly system trained only on PCA will set its Hotelling's $T^2$ threshold based on the inflated eigenvalue magnitudes. A genuine structural fault producing a small load-pathway shift in Sensors 1–3 might be dwarfed by Sensor 4's noise floor, making the fault statistically invisible. Conversely, random sensor noise spikes could trigger false alarms. The engineer has **no mechanism** to distinguish "noise large" from "structure changed".

### Question 2: Decoupling Structural Loading via Rotation

**1. How Varimax creates simple structure:**

PCA imposes a strict orthogonal hierarchy: $\text{PC}_1$ grabs maximum variance from all sensors simultaneously, $\text{PC}_2$ grabs maximum residual variance, etc. Every component loads on all sensors with non-zero weights. Varimax relaxes the hierarchy constraint and instead maximises $\sum_r \sum_j \lambda_{j,r}^4$ — the sum of fourth powers of loadings. This drives each factor toward a **sparse** loading pattern: a few sensors load near ±1, the rest load near 0. The result is "simple structure" where each factor corresponds to a physically interpretable group.

**2. Operational advantage of the rotated heatmap:**

In a raw PCA heatmap, every principal component has mixed contributions from all four sensors, making it impossible to isolate *which physical process* has changed. With Varimax:
- Factor 1 lights up Sensors 1 & 2 → a fault in this factor immediately points to the primary load path.
- Factor 2 lights up Sensor 3 → the secondary operational mode is separately tracked.
- Sensor 4's near-zero loading on both factors (and $\varphi^2 \approx 0.98$) immediately flags it as an instrumentation problem.

A plant operator can drill down to the correct physical subsystem in seconds rather than interpreting a cloud of mixed eigenvector weights.

### Question 3: Subspace Truncation Using $T^2$ and $Q$ Profiles

**1. Behaviour of Mean $Q$ Statistic as $k$ increases:**

- $k=1 \to k=2$: The Mean $Q$ drops **sharply**, because PC 2 captures the secondary structural mode ($f_2$, driving Sensor 3). Including it removes a large, coherent chunk of residual energy.
- $k=2 \to k=3$: The Mean $Q$ drops only **slightly** ("flat elbow"), because PC 3 primarily absorbs white noise from Sensor 4 — there is no new physical dimension to explain.

**2. Why the elbow at $k=2$ identifies the true physical dimensionality:**

A sharp drop in $Q$ signals that a genuine, structured mode has been added to the principal subspace. A flat elbow means subsequent components capture only noise — including them does not reduce the structured residual. The elbow criterion identifies $k=2$ as the **intrinsic dimension** of the system: exactly the two latent processes $f_1$ and $f_2$ designed into the data.

**3. Consequence of choosing $k=3$:**

By including PC 3, you force Sensor 4's instrument noise into the "clean" principal subspace. The $T^2$ statistic now amplifies random noise spikes from a faulty sensor, generating **spurious alarms**. The residual $Q$ monitor has less noise to track, masking its early-warning ability. Effectively, you contaminate your diagnostics with irrelevant variance and lose the ability to separate structural physics from sensor malfunctions.

### Question 4: Operational Trade-offs in Predictive Maintenance

**PCA Strategy ($T^2$ + $Q$):**
- $T^2$ monitors amplitude deviations within the established principal subspace.
- $Q$ monitors structural changes (new modes appearing in the residual space).
- **Weakness:** If Sensor 4 develops a noise escalation, its large variance contaminates $T^2$ through inflated leading eigenvalues. A single poorly calibrated sensor can shift the threshold of the entire monitoring system.

**FA Strategy (Rotated Factor Scores):**
- The **uniqueness vector $\boldsymbol{\varphi}^2$** acts as a dedicated **per-channel noise floor monitor**.
- When Sensor 4 loses calibration or suffers an electrical short, its uniqueness $\varphi_4^2 \to 1.0$ and its communality $h_4^2 \to 0$. The factor model automatically isolates this noise into $\boldsymbol{\Psi}$, leaving Factor 1 and Factor 2 score trajectories **unaffected**.
- Factor score monitoring is therefore **structurally isolated** from single-sensor failures.

**Verdict:** The FA strategy is significantly more **robust** against sensor calibration loss or electrical shorts. The per-sensor uniqueness metric provides an independent diagnostic channel that the PCA framework entirely lacks. In a fleet deployment, this reduces false alarms and preserves true fault sensitivity when individual instruments degrade.

# **Q1) Prediction of fraudulent transactions**

### **1. The Probability Space $(\Omega, \mathfrak{F}, P)$**

The sample space is given by $\Omega = \{\omega_1, \omega_2, \omega_3, \omega_4\}$. Since $\Omega$ is finite, we define the $\sigma$-algebra $\mathfrak{F}$ as the power set $\mathcal{P}(\Omega)$, containing $2^4 = 16$ events.
The probability measure $P$ is defined by the given mass on singletons:


$$P(\{\omega_1\}) = 0.50, \quad P(\{\omega_2\}) = 0.10, \quad P(\{\omega_3\}) = 0.20, \quad P(\{\omega_4\}) = 0.20$$


Summing these gives $\sum_{i=1}^4 P(\{\omega_i\}) = 1.00$, satisfying the axioms of probability.

### **2. The $\sigma$-algebra Generated by $Y$, $\sigma(Y)$**

The random variable $Y$ partitions the sample space into two sets:

* $Y^{-1}(\{0\}) = \{\omega_1, \omega_2\}$ (Low-risk)
* $Y^{-1}(\{1\}) = \{\omega_3, \omega_4\}$ (High-risk)

The generated $\sigma$-algebra is the collection of all pre-images of Borel sets under $Y$:


$$\sigma(Y) = \left\{ \emptyset, \{\omega_1, \omega_2\}, \{\omega_3, \omega_4\}, \Omega \right\}$$

### **3. Measurability of $E[X | Y]$**

By the measure-theoretic definition, $E[X | Y]$ is a $\sigma(Y)$-measurable function. For a function to be measurable with respect to a $\sigma$-algebra generated by a partition, it must be constant on each atom (set) of that partition.

* The atoms of $\sigma(Y)$ are $A_0 = \{\omega_1, \omega_2\}$ and $A_1 = \{\omega_3, \omega_4\}$.
* Therefore, for any $\omega \in A_0$, $E[X|Y](\omega)$ must take the same value, and similarly for $\omega \in A_1$.

### **4. Explicit Computation of $E[X | Y]$**

We compute the average of $X$ over the atoms of $\sigma(Y)$:

* **For $Y=0$:**

$$E[X | Y=0] = \frac{E[X \mathbb{1}_{\{\omega_1, \omega_2\}}]}{P(\{\omega_1, \omega_2\})} = \frac{X(\omega_1)P(\omega_1) + X(\omega_2)P(\omega_2)}{P(\omega_1) + P(\omega_2)} = \frac{0(0.50) + 1(0.10)}{0.60} = \frac{1}{6}$$


* **For $Y=1$:**

$$E[X | Y=1] = \frac{E[X \mathbb{1}_{\{\omega_3, \omega_4\}}]}{P(\{\omega_3, \omega_4\})} = \frac{X(\omega_3)P(\omega_3) + X(\omega_4)P(\omega_4)}{P(\omega_3) + P(\omega_4)} = \frac{0(0.20) + 1(0.20)}{0.40} = \frac{1}{2}$$



Thus: $E[X | Y] = \frac{1}{6} \mathbb{1}_{\{Y=0\}} + \frac{1}{2} \mathbb{1}_{\{Y=1\}}$

### **5. Verifying the Defining Property**

We check $\int_A E[X | Y] \, dP = \int_A X \, dP$ for $A \in \sigma(Y)$:

* For $A = \{\omega_1, \omega_2\}$: $\frac{1}{6}(0.60) = 0.10$. Also $\int_A X \, dP = 0(0.5) + 1(0.1) = 0.10$.
* For $A = \{\omega_3, \omega_4\}$: $\frac{1}{2}(0.40) = 0.20$. Also $\int_A X \, dP = 0(0.2) + 1(0.2) = 0.20$.
The property holds.

### **6. Conditional PMF $p_{X|Y}(x | y)$**

Using $P(X=x | Y=y) = \frac{P(X=x, Y=y)}{P(Y=y)}$:

* **$y=0$**: $p_{X|Y}(1|0) = \frac{0.1}{0.6} = \frac{1}{6}$, $p_{X|Y}(0|0) = \frac{0.5}{0.6} = \frac{5}{6}$
* **$y=1$**: $p_{X|Y}(1|1) = \frac{0.2}{0.4} = \frac{1}{2}$, $p_{X|Y}(0|1) = \frac{0.2}{0.4} = \frac{1}{2}$

### **7. Identity Verification**

Check $E[X | Y=y] = \sum_{x \in \{0,1\}} x \cdot p_{X|Y}(x | y)$:

* For $y=0$: $0(\frac{5}{6}) + 1(\frac{1}{6}) = \frac{1}{6}$ (Matches Part 4)
* For $y=1$: $0(\frac{1}{2}) + 1(\frac{1}{2}) = \frac{1}{2}$ (Matches Part 4)

### **8. Mean Square Error (MSE)**

$$MSE = E[(X - E[X|Y])^2] = \sum_{i=1}^4 (X(\omega_i) - E[X|Y](\omega_i))^2 P(\omega_i)$$

$$MSE = (0-\frac{1}{6})^2(0.5) + (1-\frac{1}{6})^2(0.1) + (0-\frac{1}{2})^2(0.2) + (1-\frac{1}{2})^2(0.2)$$

$$MSE = \frac{0.5}{36} + \frac{2.5}{36} + 0.05 + 0.05 = \frac{3}{36} + 0.1 = \frac{1}{12} + \frac{1}{10} = \frac{11}{60} \approx 0.1833$$

### **9. Comparison with Constant Estimate $E[X]$**

If we ignore $Y$, our best estimate is $E[X] = 0.30$. The error is the variance:


$$Var(X) = E[X^2] - (E[X])^2 = 0.30 - (0.3)^2 = 0.30 - 0.09 = 0.21$$


The MSE reduced from $0.21$ to $0.1833$. By the **Law of Total Variance**:


$$Var(X) = E[Var(X|Y)] + Var(E[X|Y])$$


The information $Y$ "captures" some of the variance of $X$ ($Var(E[X|Y])$), leaving a smaller residual error ($E[Var(X|Y)]$).

### **10. Information Theory Analysis**

* **a) Entropy $H(X)$:**

$$H(X) = -[0.7 \log_2(0.7) + 0.3 \log_2(0.3)] \approx 0.8813 \text{ bits}$$


* **b) Conditional Entropy $H(X|Y)$:**

$$H(X|Y) = P(Y=0)H(X|Y=0) + P(Y=1)H(X|Y=1)$$


$$H(X|Y) = 0.6 \left( -\frac{5}{6}\log_2\frac{5}{6} - \frac{1}{6}\log_2\frac{1}{6} \right) + 0.4(1) \approx 0.6(0.65) + 0.4 = 0.79 \text{ bits}$$


* **c) Mutual Information $I(X;Y)$:**

$$I(X;Y) = H(X) - H(X|Y) = 0.8813 - 0.79 = 0.0913 \text{ bits}$$


* **d) Percentage removed:**

$$\frac{I(X;Y)}{H(X)} \times 100 \approx \frac{0.0913}{0.8813} \times 100 \approx 10.36\%$$

# **Q2) Conditional Density of the Multivariate Gaussian**
To show that the conditional distribution of $X$ given $Y=y$ for a multivariate Gaussian is also Gaussian with the specified mean and covariance, we use the definition of conditional density:


$$f_{X|Y}(x|y) = \frac{f_{X,Y}(x,y)}{f_Y(y)}$$

### 1. The Marginal Density of $Y$

Since $Z = [X, Y]^T$ is multivariate normal, its marginal distributions are also normal. Specifically, $Y \sim \mathcal{N}(\mu_Y, \Sigma_{YY})$:


$$f_Y(y) = \frac{1}{(2\pi)^{m/2} |\Sigma_{YY}|^{1/2}} \exp\left( -\frac{1}{2} (y-\mu_Y)^T \Sigma_{YY}^{-1} (y-\mu_Y) \right)$$

### 2. Block Matrix Inversion and the Schur Complement

To simplify the joint density $f_{X,Y}(z)$, we use the block inversion formula for $\Sigma$. Let $S_{YY} = \Sigma_{YY}$ and define the **Schur Complement** of $\Sigma_{YY}$ in $\Sigma$ as:


$$\Sigma_{X|Y} = \Sigma_{XX} - \Sigma_{XY}\Sigma_{YY}^{-1}\Sigma_{YX}$$


The inverse of the partitioned covariance matrix $\Sigma^{-1}$ can be written as:


$$\Sigma^{-1} = \begin{bmatrix} I & 0 \\ -\Sigma_{YY}^{-1}\Sigma_{YX} & I \end{bmatrix}^T \begin{bmatrix} \Sigma_{X|Y}^{-1} & 0 \\ 0 & \Sigma_{YY}^{-1} \end{bmatrix} \begin{bmatrix} I & -\Sigma_{XY}\Sigma_{YY}^{-1} \\ 0 & I \end{bmatrix}$$


Also, the determinant property states: $|\Sigma| = |\Sigma_{YY}| \cdot |\Sigma_{X|Y}|$.

### 3. Decomposing the Quadratic Form

The exponent in the joint density is $Q = (z-\mu)^T \Sigma^{-1} (z-\mu)$. Let $u = x - \mu_X$ and $v = y - \mu_Y$. Using the decomposition of $\Sigma^{-1}$:


$$Q = (u - \Sigma_{XY}\Sigma_{YY}^{-1}v)^T \Sigma_{X|Y}^{-1} (u - \Sigma_{XY}\Sigma_{YY}^{-1}v) + v^T \Sigma_{YY}^{-1} v$$


Substitute back $u$ and $v$:


$$Q = (x - (\mu_X + \Sigma_{XY}\Sigma_{YY}^{-1}(y-\mu_Y)))^T \Sigma_{X|Y}^{-1} (x - (\mu_X + \Sigma_{XY}\Sigma_{YY}^{-1}(y-\mu_Y))) + (y-\mu_Y)^T \Sigma_{YY}^{-1} (y-\mu_Y)$$


Let $\mu^* = \mu_X + \Sigma_{XY}\Sigma_{YY}^{-1}(y-\mu_Y)$. The quadratic form becomes:


$$Q = (x - \mu^*)^T \Sigma_{X|Y}^{-1} (x - \mu^*) + (y-\mu_Y)^T \Sigma_{YY}^{-1} (y-\mu_Y)$$

### 4. Combining terms for $f_{X|Y}(x|y)$

Now substitute the joint and marginal densities into the conditional density formula:


$$f_{X|Y}(x|y) = \frac{\frac{1}{(2\pi)^{(n+m)/2} |\Sigma|^{1/2}} \exp\left( -\frac{1}{2} [ (x-\mu^*)^T \Sigma_{X|Y}^{-1} (x-\mu^*) + (y-\mu_Y)^T \Sigma_{YY}^{-1} (y-\mu_Y) ] \right)}{\frac{1}{(2\pi)^{m/2} |\Sigma_{YY}|^{1/2}} \exp\left( -\frac{1}{2} (y-\mu_Y)^T \Sigma_{YY}^{-1} (y-\mu_Y) \right)}$$

Cancel the terms involving $y$ in the exponent and simplify the constants:

1. **Constants:** $\frac{(2\pi)^{m/2}}{(2\pi)^{(n+m)/2}} = \frac{1}{(2\pi)^{n/2}}$
2. **Determinants:** $\frac{|\Sigma_{YY}|^{1/2}}{|\Sigma|^{1/2}} = \frac{|\Sigma_{YY}|^{1/2}}{(|\Sigma_{YY}| \cdot |\Sigma_{X|Y}|)^{1/2}} = \frac{1}{|\Sigma_{X|Y}|^{1/2}}$

### 5. Final Result

The resulting density is:


$$f_{X|Y}(x|y) = \frac{1}{(2\pi)^{n/2} |\Sigma_{X|Y}|^{1/2}} \exp\left( -\frac{1}{2} (x - \mu^*)^T \Sigma_{X|Y}^{-1} (x - \mu^*) \right)$$


Where:


$$\Sigma_{X|Y} = \Sigma_{XX} - \Sigma_{XY}\Sigma_{YY}^{-1}\Sigma_{YX}$$

$$\mu^* = \mu_X + \Sigma_{XY}\Sigma_{YY}^{-1}(y - \mu_Y)$$


This identifies the conditional distribution as $X|Y=y \sim \mathcal{N}(\mu^*, \Sigma_{X|Y})$.

# Q3) Continuous Fault Monitoring of a Spindle Motor
### **1. Joint Distribution**

Since $X$ and $Q$ are independent normal random variables, any linear combination of them is normal. $Y = X + Q$ is such a linear combination, making $(X, Y)$ a multivariate normal vector.

* **Mean of $Y$:** $E[Y] = E[X + Q] = E[X] + E[Q] = 0 + 0 = 0$.
* **Variance of $Y$:** Since $X$ and $Q$ are independent, $Var(Y) = Var(X) + Var(Q) = 1 + 4 = 5$. Thus, $Y \sim N(0, 5)$.
* **Covariance $Cov(X, Y)$:**

$$Cov(X, Y) = Cov(X, X + Q) = Cov(X, X) + Cov(X, Q)$$



Since $X \perp Q$, $Cov(X, Q) = 0$. Therefore, $Cov(X, Y) = Var(X) = 1$.
* **Joint Distribution:**

$$\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathcal{N} \left( \begin{bmatrix} 0 \\ 0 \end{bmatrix}, \begin{bmatrix} 1 & 1 \\ 1 & 5 \end{bmatrix} \right)$$



### **2. Information Geometry**

$X$ is not $\sigma(Y)$-measurable because there is no Borel function $g$ such that $X = g(Y)$ almost surely. In the $\sigma$-algebra sense, $\sigma(Y) \subset \sigma(X, Q)$, and $X$ is not contained in the sub-$\sigma$-algebra generated by their sum.

**Practical Interpretation:** If $Y=5$, the equation $5 = X + Q$ has infinitely many solutions. One cannot distinguish between a "hot motor" ($X=4, Q=1$) and "hard material" ($X=0, Q=5$). The observation $Y$ conflates the two sources of variance; hence, $X$ remains a random variable even after $Y$ is known.

### **3. Conditional Expectation (The Best Estimate)**

Using the formula for joint normals:


$$E[X|Y=y] = \mu_X + \frac{Cov(X, Y)}{Var(Y)}(y - \mu_Y) = 0 + \frac{1}{5}(y - 0) = \frac{1}{5}y$$


**Signal-to-Total-Variance Ratio:**


$$E[X|Y] = \left( \frac{\sigma_X^2}{\sigma_X^2 + \sigma_Q^2} \right) Y$$


Here, the "signal" we want to estimate is $X$ (variance 1), and the "total variance" of the observation is 5. We weight our observation $Y$ by the proportion of its variance that is actually attributed to $X$.

### **4. Verification of the Residual**

Define $Z = X - \hat{X}$ where $\hat{X} = \frac{1}{5}Y$.

* **Mean of $Z$:** $E[Z] = E[X] - \frac{1}{5}E[Y] = 0 - 0 = 0$.
* **Orthogonality Principle:**

$$Cov(Z, Y) = E[ZY] - E[Z]E[Y] = E[(X - \frac{1}{5}Y)Y] = E[XY] - \frac{1}{5}E[Y^2]$$



Recall $E[XY] = Cov(X,Y) = 1$ and $E[Y^2] = Var(Y) = 5$.

$$Cov(Z, Y) = 1 - \frac{1}{5}(5) = 1 - 1 = 0$$



Since they are joint normal and uncorrelated, the error $Z$ is independent of the observation $Y$.

### **5. Information Theory: Differential Entropy**

The differential entropy of a normal distribution $N(\mu, \sigma^2)$ is $h = \frac{1}{2}\ln(2\pi e \sigma^2)$.

* **Entropy of $X$:** $h(X) = \frac{1}{2}\ln(2\pi e (1)) = \frac{1}{2}\ln(2\pi e)$.
* **Conditional Entropy $h(X|Y)$:** We use the conditional variance $\sigma^2_{X|Y} = \sigma^2_X - \frac{Cov(X,Y)^2}{Var(Y)} = 1 - \frac{1}{5} = 0.8$.

$$h(X|Y) = \frac{1}{2}\ln(2\pi e (0.8))$$



### **6. Mutual Information**

$$I(X; Y) = h(X) - h(X|Y) = \frac{1}{2}\ln\left(\frac{2\pi e}{2\pi e \cdot 0.8}\right) = \frac{1}{2}\ln\left(\frac{1}{0.8}\right) = \frac{1}{2}\ln(1.25) \approx 0.1116 \text{ nats.}$$

* **Uncertainty Reduction:** The ratio of variances is $0.8/1.0$. The percentage of entropy reduction is not linear with variance, but we can look at the reduction in "information bits": $\frac{h(X) - h(X|Y)}{h(X)}$.
* **If $Var(Q) = 100$:** $Var(Y) = 101$. $Cov(X,Y) = 1$. $\sigma^2_{X|Y} = 1 - \frac{1}{101} \approx 0.99$. $I(X;Y) = \frac{1}{2}\ln(\frac{1}{0.99})$, which is nearly 0. As noise increases, $Y$ provides almost no information about $X$.

### **7. Diagnostic Interpretation**

The manager is incorrect because $Y$ is a "noisy" version of $X$.

1. **Bias:** While $E[Y|X] = X$ (unbiased sensing), the manager wants to go the other way. $E[X|Y] = 0.2Y$. Using $Y$ alone as an estimator for $X$ results in an estimator that overestimates the magnitude of the temperature because it fails to account for the mechanical noise $Q$.
2. **Shrinkage:** The factor $\frac{1}{5}$ is a "shrinkage estimator." Because we know $X$ usually sits near its mean (0), and we know $Y$ is contaminated by noise, the most probable value for $X$ given a high $Y$ is a value "pulled" back toward the prior mean. This accounts for the fact that a large $Y$ is likely caused by *both* a somewhat high $X$ and a somewhat high $Q$ simultaneously.
